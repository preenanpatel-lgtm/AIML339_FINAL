## MANUAL QUERIES

Test five ingredient groups with 3, 4, 5 and 6 ingredients.

```python
# TEST FIXED MANUAL QUERIES

# create 20 manual queries - 5 groups with 3, 4, 5 and 6 ingredients
manual_queries = make_manual_queries()
manual_baseline_rows = []

for _, query_row in manual_queries.iterrows():
    # compare all three baseline methods using the same query
    for model_name, model_function in {"TF-IDF": tfidf_model, "Jaccard": jaccard_model, "MiniLM": minilm_model}.items():
        recommendations = model_function(query_row["query"], top_k=5)

        # save the top 5 recipes for each query and baseline
        for rank, (_, recipe) in enumerate(recommendations.iterrows(), start=1):
            manual_baseline_rows.append({"query_id": query_row["query_id"], "group": query_row["group"], "query_length": query_row["query_length"], "query": query_row["query"], "model": model_name, "rank": rank, "recipe_title": recipe["recipe_title"], "recipe_ingredients": ", ".join(recipe["processing_ingredients"])})

manual_baseline_recommendations = pd.DataFrame(manual_baseline_rows)
manual_baseline_recommendations.to_csv("baseline_manual_recommendations.csv",
    index=False)
```
